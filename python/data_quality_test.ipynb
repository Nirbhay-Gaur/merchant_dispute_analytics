```python
## DATA QUALITY TESTS
```

```python
checks = {
    "duplicate_transaction_ids": df_transactions["transaction_id"].duplicated().sum(),
    "null_transaction_amounts": df_transactions["amount"].isnull().sum(),
    "null_dispute_amounts": df_disputes['amount'].isna().sum(),
    "negative_transaction_amounts": (df_transactions["amount"] <= 0).sum(),
    "negative_dispute_amounts": (df_disputes["amount"] <= 0).sum(),
    "negative_response_days": (df_disputes["response_days"] < 0).sum(),
    "invalid_merchant_won": (~df_disputes["merchant_won"].isin([0, 1])).sum(),
    "orphaned_disputes": (~df_disputes["transaction_id"].isin(df_transactions["transaction_id"])).sum(),
    "dispute_date_out_of_range": (~df_disputes['transaction_date'].between('2025-01-01', '2025-12-31')).sum(),
}
for check, result in checks.items():
    status = "PASS" if result == 0 else "FAIL"
    print(f"[{status}] {check}: {result}")
```
